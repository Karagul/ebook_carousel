```python
from PIL import Image, ImageDraw, ImageFont
import os

# --- SETTINGS ---
output_dir = "/Users/huseyinkaragul/Jupyter_Notebooks/16_Document_Generation"
card_width = 1000   # wider horizontal layout
card_height = 600   # shorter height
padding = 40
bg_color = "white"
title_font_size = 36
prompt_font_size = 24
footer_font_size = 20
line_spacing = 10
max_prompt_lines = 10  # max to prevent overflow

# --- FONT SETUP ---
try:
    font_path = "/Library/Fonts/Arial Unicode.ttf"
    title_font = ImageFont.truetype(font_path, title_font_size)
    prompt_font = ImageFont.truetype(font_path, prompt_font_size)
    footer_font = ImageFont.truetype(font_path, footer_font_size)
except:
    title_font = ImageFont.load_default()
    prompt_font = ImageFont.load_default()
    footer_font = ImageFont.load_default()

# --- HELPERS ---
def clean(text):
    return (text.replace("’", "'").replace("“", '"').replace("”", '"')
                .replace("–", "-").replace("—", "-"))

def wrap_text(draw, text, font, max_width, max_lines=None):
    words = clean(text).split()
    lines, line = [], ""
    for word in words:
        test_line = f"{line} {word}".strip()
        if draw.textbbox((0, 0), test_line, font=font)[2] > max_width:
            lines.append(line.strip())
            line = word
            if max_lines and len(lines) >= max_lines:
                lines[-1] += " ..."
                break
        else:
            line = test_line
    if len(lines) < (max_lines or 999):
        lines.append(line.strip())
    return lines

# --- PROMPT CARDS ---
cards = [
    {
        "title": "Discover Profitable Topics",
        "prompt": "Suggest 10 trending, low-competition hot ebook topics. They should be ideal for Amazon KDP, Gumroad, or Etsy and fit well with Canva-style templates.",
        "tagline": "Choose a profitable idea from the start!"
    },
    {
        "title": "Generate Click-Worthy Titles",
        "prompt": "I'm writing an ebook on: [insert topic]. Suggest 10 attention-grabbing titles that drive clicks and boost discoverability.",
        "tagline": "Titles too irresistible to scroll past"
    },
    {
        "title": "Write an Addictive Chapter",
        "prompt": "Write a chapter that's simple, fluent, and full of storytelling. It should keep the reader engaged from beginning to end.",
        "tagline": "Content that creates addiction"
    },
    {
        "title": "Design Like a Pro (No Designer)",
        "prompt": "Suggest 5 Canva layout templates that match the vibe of my ebook. Include font pairings, color palettes, and layout tips to give it a premium look without hiring a designer.",
        "tagline": "Look premium without hiring a designer"
    },
    {
        "title": "Build a Sales Page That Prints Money",
        "prompt": "Write a high-converting sales page for my ebook on [insert topic]. Include a strong headline, emotional benefits, short testimonials, and an irresistible call to action (CTA).",
        "tagline": "Write a page that actually sells!"
    },
    {
        "title": "Plan to Sell First 1,000 Copies",
        "prompt": "Create a step-by-step campaign plan to sell 1,000 copies of my ebook using social media, email lists, and strategic partnerships — all using free traffic sources.",
        "tagline": "Your roadmap to the first 1,000 sales"
    },
    {
        "title": "Scale into a 6-Figure Brand",
        "prompt": "Create a growth plan to turn my ebook into a 6-figure brand. Include steps like product expansion, bundling, partnerships, and scaling with paid ads.",
        "tagline": "Begin your transformation into a brand"
    }
]

# --- RENDER EACH CARD ---
for i, card in enumerate(cards):
    img = Image.new("RGB", (card_width, card_height), color=bg_color)
    draw = ImageDraw.Draw(img)

    # Title
    draw.text((padding, padding), clean(card["title"]), font=title_font, fill="black")

    # Prompt: horizontal block to the right of title
    prompt_max_width = card_width - 2 * padding
    prompt_lines = wrap_text(draw, card["prompt"], prompt_font, prompt_max_width, max_lines=max_prompt_lines)

    y_start = padding + 70
    for line in prompt_lines:
        draw.text((padding, y_start), line, font=prompt_font, fill="black")
        y_start += prompt_font_size + line_spacing

    # Footer
    draw.text((padding, card_height - 80), clean(card["tagline"]), font=footer_font, fill="darkred")
    draw.text((padding, card_height - 50), "AMARANTH BROSE", font=footer_font, fill="grey")

    # Save as individual horizontal panel
    filename = f"swipe_card_{i+1}.png"
    img.save(os.path.join(output_dir, filename))
    print(f"✅ Saved: {filename}")
```
